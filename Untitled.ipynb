```python
import pandas as pd

# Convert columns to datetime
orders["order_delivered_timestamp"] = pd.to_datetime(
    orders["order_delivered_timestamp"],
    errors="coerce"
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"],
    errors="coerce"
)

# Create delay_days in orders
orders["delay_days"] = (
    orders["order_delivered_timestamp"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Recreate delivered AFTER creating delay_days
delivered = orders[
    orders["order_delivered_timestamp"].notna()
].copy()

# Keep rows with valid delay information
valid_deliveries = delivered[
    delivered["delay_days"].notna()
].copy()

# Calculate on-time and late orders
on_time_orders = (
    valid_deliveries["delay_days"] <= 0
).sum()

late_orders = (
    valid_deliveries["delay_days"] > 0
).sum()

total_valid_deliveries = len(valid_deliveries)

# Calculate rates
on_time_rate = (
    on_time_orders / total_valid_deliveries
) * 100

late_delivery_rate = (
    late_orders / total_valid_deliveries
) * 100

print("Valid deliveries:", total_valid_deliveries)
print("On-time orders:", on_time_orders)
print("Late orders:", late_orders)
print("On-time delivery rate:", round(on_time_rate, 2), "%")
print("Late-delivery rate:", round(late_delivery_rate, 2), "%")
```

```python
# Keep delivered orders that have a valid delivery-delay value
valid_deliveries = orders[
    (orders["order_status"] == "delivered") &
    (orders["delivery_delay_days"].notna())
]

# Count late orders
late_orders = (
    valid_deliveries["delivery_delay_days"] > 0
).sum()

# Calculate late delivery rate
late_rate = (
    late_orders / len(valid_deliveries)
) * 100

print("Late Orders:", late_orders)
print("Total Valid Deliveries:", len(valid_deliveries))
print("Late Delivery Rate:", round(late_rate, 2), "%")
```
